#### 1. Create a class called `Book` that has the attributes title, author, pages, price. It also has the method get_price() which returns price, and set_price() to ensure price is numeric.

Then create two instances of this class:

b1:
* title -> 'Practical Programming'
* author -> 'Gries, Campbell, Montojo'
* pages -> 383
* price -> 50

b2:
* title -> 'Building a Career in Data Science'
* author -> 'Robinson, Nolis'
* pages -> 322
* price -> 40

```python
class Book:
  def __init__(self, title, author, pages, price):
    self.title = title
    self.author = author
    self.pages = pages
    self.price = price
  def get_price(self):
        return self.price
  def set_price(self, new_price):
        if type(new_price) == int or type(new_price) == float:
            self.price = new_price
        else:
            print("Price must be numeric")
b1 = Book('Practical Programming','Gries, Campbell, Montojo', 383, 50)
b2 = Book('Building a Career in Data Science', 'Robinson, Nolis', 322, 40)
print(b1.get_price())
print(b2.get_price())
```

#### 2. Create a `Phone` class that has the attributes model, id, name, brand and price. Create the getters and setters for each attribute. As well as the following methods: calculate_total which takes the price and calculates it by the tax argument.

```python
class Phone:
  def __init__(self, model, id, name, brand, price):
        self.model = model
        self.id = id
        self.name = name
        self.brand = brand
        self.price = price
  def get_model(self):
      return self.model
  def get_id(self):
      return self.id
  def get_name(self):
      return self.name
  def get_brand(self):
      return self.brand
  def get_price(self):
      return self.price

  def calculate_total(self, tax):
      return self.price + (self.price * tax)

  phone1 = Phone('iPhone 16', 101, 'Ellie Phone', 'Apple', 1200)

  print(phone1.calculate_total(0.13))
```

#### 1. Write a code to convert a 1D array to a 2D array with 2 rows.

Original Array: [0 1 2 3 4 5]

Reshaped 2x3 Array:

[[0 1 2]

[3 4 5]]

#### 4. Reverse a 1D NumPy array
Ex. arr = np.arange(10)

Output is [9 8 7 6 5 4 3 2 1 0]

*Hint! Use slicing with step = -1.*

```python
import numpy as np
arr = np.arange(10)
reversed_arr = arr[::-1]
print(reversed_arr)
```

#### 5. Stack these arrays horizontally


a = np.array([1, 2, 3])

b = np.array([4, 5, 6])

```python
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
stacked = np.hstack((a, b))
print(stacked)
```
